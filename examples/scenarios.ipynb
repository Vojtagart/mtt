```python
import numpy as np
from typing import Optional
%matplotlib widget
import matplotlib.pyplot as plt
import imageio
from IPython.display import Image, display
from matplotlib.patches import Circle

import mtt
```

# Plotting function

```python
def scenario_to_gif(fig: plt.Figure, tmap: mtt.TrajMap, plotter: mtt.Plotter, path_name: str, fps: float = 5.0):
    """
    Simulates the tracking scenario and saves the output directly to a GIF

    Parameters:
    - fig: Matplotlib Figure associated with the plotter
    - tmap: TrajMap instance containing the generated trajectory timeline
    - plotter: Plotter instance for visualization
    - path_name: Path to save the GIF
    - fps: Frames per second for the resulting GIF
    """
    frames = []
    ax = plotter.ax

    ax.add_patch(Circle(tmap.center, tmap.rad, facecolor='none', edgecolor='gray', ls='--'))

    for t, trajs in enumerate(tmap):
        ax.set_title(f"Time {t + 1}")

        plotter.plot_trajs(t, trajs)
        ax.legend(loc='upper right')

        fig.canvas.draw()
        frame = np.array(fig.canvas.renderer.buffer_rgba())
        frames.append(frame)
    
    imageio.mimsave(path_name, frames, fps=fps, loop=1)
    plt.close('all')
```

```python
config = {'uest_traj': False, 'ucov_ellipse': False, 'est_traj': False, 'target': False, 'pregen_traj': False,
          'traj_arrow': True, 'traj_meas': True, 'cov_ellipse': False, 'clutter': False, 'cur_dets': False}
```

# One-Crossing scenario

```python
rng = mtt.get_rng(seed=44)
tmap = mtt.scenario_one_crossing(ndat=100, ntargets=5, radius=100., PS=0.995, q=0.02, r=1., rng=rng)

fig, ax = plt.subplots(figsize=(6, 6))
plotter = mtt.Plotter(ax, params={
            'traj_arrow': {'closed':True, 'facecolor':'black', 'edgecolor':'none', 'zorder':10, 'arrow_scale':3.},
            'traj_meas': {'color':'blue', 'marker':'.', 'label':'measurements', 's':8, 'alpha':0.6},
        },
        config = {'uest_traj': False, 'ucov_ellipse': False, 'est_traj': False, 'target': False, 'pregen_traj': False,
                  'traj_arrow': True, 'traj_meas': True, 'cov_ellipse': False, 'clutter': False, 'cur_dets': False}
)

path_name = "gifs/one_cross.gif"
scenario_to_gif(fig, tmap, plotter, path_name=path_name)
display(Image(filename=path_name))
```
